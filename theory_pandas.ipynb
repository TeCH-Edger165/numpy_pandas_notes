```python
# Group by:
l=[['A','S1',100,10],['A','S2',200,25],['B','S1',120,15],['A','S2',150,20]]
c=['Category','Store','Sales','Quantity']
df=pd.DataFrame(l,columns=c)
df
```

```python
# Total sum
df.sum()[['Sales','Quantity']]
```

```python
# Category wise distribution
cat=df.groupby('Category')
for i,v in cat:
    print(i)
    print(v)
    print("-----------")
```

```python
# Category wise sum:
cat.sum()[['Sales','Quantity']]
```

```python
# Grouping Multiple:
cat2=df.groupby(['Category','Store'])
for i,v in cat2:
    print(i)
    print(v)
```

```python
# Aggregation Functions::
df['Sales'].mean()
```

```python
df['Sales'].agg(['mean','median','std','count'])
```
